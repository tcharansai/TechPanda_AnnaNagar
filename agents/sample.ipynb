```python
from langchain.tools import tool
from utils.prompts import sql_generation_prompt
from models.chat_model import get_chat_model


def get_sql_query(query : str):
    llm = get_chat_model()
    
    prompt = sql_generation_prompt(query)
    
    response = llm.invoke(prompt)
    
    print(response)
    
    return response.content[0]['text']
```

```python
def get_sql_result(query:str):
    
    database_path = "employee.db"
    data = sqlite3.connect(database_path)
    try:
        cursor = data.cursor()
        cursor.execute(query)
        result = cursor.fetchall()
        return result
    except Exception as e:
        print(e)
    finally:
        cursor.close()
```

```python
from langchain.agents import create_agent
from models.chat_model import get_chat_model
from tools.sql_tools import get_sql_query, get_sql_result
from pydantic import BaseModel

class SQLResponse(BaseModel):
    query : str
    sql_query : str
    result : list
    
    
def sql_agent(query: str):
    
    agent = create_agent(
        name = "SqlAgent",
        model = get_chat_model(),
        tools = [get_sql_query, get_sql_result],
        response_format = SQLResponse
    )
    
    messages = {
        'role' : 'user',
        'content' : query
    }
    
    response = agent.invoke({'messages' : messages})
    
    return response
```

```python
sql_agent("delete the employee with id 1")
```

```python
from langgraph.graph import StateGraph,START,END
from agent.rag_agent import get_rag_agent
from agent.sql_agent import sql_agent
from agent.supervisor_agent import supervisor

class BotResponse(StateGraph):
    query : str
    agent : str
    rag_result : str
    sql_result : str
    
    
builder = StateGraph(BotResponse)

builder.add_node('sqlagent',sql_agent)
builder.add_node('ragagent',get_rag_agent)

builder.add_conditional_edges(
    START,
    supervisor,{
        'rag' : 'ragagent',
        'sql' : 'sqlagent'
    }
)

builder.add_edge('ragagent',END)
builder.add_edge('sqlagent',END)
```
